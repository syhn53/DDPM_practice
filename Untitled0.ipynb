```python
# 1. 필요한 라이브러리 불러오기
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn

# 2. 실행 환경 확인
device = "cuda" if torch.cuda.is_available() else "cpu"

print("PyTorch 버전:", torch.__version__)
print("실행 장치:", device)
```

```python
# 1. 숫자 이미지 데이터셋 불러오기
from sklearn.datasets import load_digits

digits = load_digits()

# 2. 이미지와 정답 확인
images = digits.images
labels = digits.target

print("이미지 데이터 모양:", images.shape)
print("정답 데이터 모양:", labels.shape)
print("첫 번째 이미지의 정답:", labels[0])
```

```python
# 1. 원본 이미지 하나 선택
x_start = images[0].astype(np.float32)

# 2. 픽셀값을 0~1 범위로 정규화
x_start = x_start / 16.0

# 3. 가우시안 노이즈 생성
noise = np.random.randn(8, 8).astype(np.float32)

# 4. 노이즈를 추가하는 정도 설정
noise_levels = [0.0, 0.2, 0.5, 0.8, 1.0]

# 5. 결과 시각화
fig, axes = plt.subplots(1, 5, figsize=(12, 3))

for ax, level in zip(axes, noise_levels):
    x_noisy = x_start + level * noise

    ax.imshow(x_noisy, cmap="gray", vmin=-1, vmax=1)
    ax.set_title(f"Noise: {level}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
# 1. 전체 노이즈 추가 단계 설정
T = 100

# 2. 각 단계의 beta 값 생성
betas = torch.linspace(0.0001, 0.02, T)

# 3. alpha와 누적 alpha 계산
alphas = 1.0 - betas
alphas_cumprod = torch.cumprod(alphas, dim=0)

# 4. 원본 이미지 준비
x0 = torch.tensor(images[0], dtype=torch.float32) / 16.0

# 5. 가우시안 노이즈 생성
noise = torch.randn_like(x0)

# 6. 여러 단계에서 노이즈 추가 결과 확인
timesteps = [0, 9, 29, 59, 99]

fig, axes = plt.subplots(1, len(timesteps) + 1, figsize=(14, 3))

axes[0].imshow(x0, cmap="gray", vmin=-1, vmax=1)
axes[0].set_title("Original")
axes[0].axis("off")

for ax, t in zip(axes[1:], timesteps):
    x_t = (
        torch.sqrt(alphas_cumprod[t]) * x0
        + torch.sqrt(1 - alphas_cumprod[t]) * noise
    )

    ax.imshow(x_t, cmap="gray", vmin=-1, vmax=1)
    ax.set_title(f"t = {t + 1}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
# 필요한 라이브러리
from torch.utils.data import DataLoader, TensorDataset

# 이미지 데이터를 PyTorch 텐서로 변환
x_data = torch.tensor(
    images / 16.0,
    dtype=torch.float32
).unsqueeze(1)

# 데이터셋과 데이터 로더 만들기
dataset = TensorDataset(x_data)
loader = DataLoader(
    dataset,
    batch_size=64,
    shuffle=True
)

# 노이즈와 시간 t를 입력받는 신경망
class NoisePredictor(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 1, kernel_size=3, padding=1)
        )

        # 시간 정보를 입력받아 각 픽셀에 반영
        self.time_embed = nn.Embedding(T, 8)
        self.time_proj = nn.Linear(8, 1)

    def forward(self, x, t):
        time_info = self.time_proj(
            self.time_embed(t)
        ).view(-1, 1, 1, 1)

        return self.net(x + time_info)

model = NoisePredictor().to(device)

print("이미지 데이터 모양:", x_data.shape)
print("모델 준비 완료!")
```

```python
# Adam 최적화기
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)

# 학습에 사용할 beta, alpha 계수를 GPU/CPU 장치에 준비
betas = betas.to(device)
alphas = alphas.to(device)
alphas_cumprod = alphas_cumprod.to(device)

model.train()

epochs = 20
loss_history = []

for epoch in range(epochs):
    total_loss = 0.0

    for (x0,) in loader:
        x0 = x0.to(device)
        batch_size = x0.shape[0]

        # 이미지마다 무작위 시간 단계 선택
        t = torch.randint(
            0, T, (batch_size,), device=device
        )

        # 가우시안 노이즈 생성
        noise = torch.randn_like(x0)

        # 논문의 정방향 수식으로 노이즈 이미지 생성
        alpha_bar = alphas_cumprod[t].view(
            batch_size, 1, 1, 1
        )

        x_t = (
            torch.sqrt(alpha_bar) * x0
            + torch.sqrt(1 - alpha_bar) * noise
        )

        # 신경망이 노이즈 예측
        predicted_noise = model(x_t, t)

        # 실제 노이즈와 예측 노이즈의 차이
        loss = nn.functional.mse_loss(
            predicted_noise, noise
        )

        # 신경망 업데이트
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(loader)
    loss_history.append(avg_loss)

    print(
        f"Epoch {epoch + 1:02d}/{epochs}, "
        f"Loss: {avg_loss:.4f}"
    )
```

```python
# 모델을 평가 모드로 변경
model.eval()

# 생성할 이미지 개수
num_samples = 10

# 무작위 노이즈에서 시작
x = torch.randn(num_samples, 1, 8, 8)

# 이전 시점의 누적 알파값 계산
alpha_bar_prev = torch.cat([
    torch.tensor([1.0]),
    alphas_cumprod[:-1]
])

# 역방향 과정의 분산 계산
posterior_variance = (
    betas * (1 - alpha_bar_prev)
    / (1 - alphas_cumprod)
)

# 노이즈를 단계적으로 제거
with torch.no_grad():
    for t in reversed(range(T)):
        t_batch = torch.full(
            (num_samples,), t, dtype=torch.long
        )

        # 현재 이미지의 노이즈 예측
        predicted_noise = model(x, t_batch)

        beta_t = betas[t]
        alpha_t = alphas[t]
        alpha_bar_t = alphas_cumprod[t]

        # 이전 시점 이미지의 평균 계산
        mean = (
            x - beta_t / torch.sqrt(1 - alpha_bar_t)
            * predicted_noise
        ) / torch.sqrt(alpha_t)

        # 마지막 단계가 아니면 무작위 노이즈 추가
        if t > 0:
            x = (
                mean
                + torch.sqrt(posterior_variance[t])
                * torch.randn_like(x)
            )
        else:
            x = mean

# 이미지 형태로 변환
samples = x.squeeze(1).numpy()

# 생성된 이미지 10장 출력
fig, axes = plt.subplots(1, 10, figsize=(15, 2))

for i, ax in enumerate(axes):
    ax.imshow(samples[i], cmap="gray", vmin=0, vmax=1)
    ax.axis("off")
    ax.set_title(str(i + 1))

plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class NoisePredictor(nn.Module):
    def __init__(self):
        super().__init__()

        self.time_embed = nn.Embedding(T, 16)

        self.net = nn.Sequential(
            nn.Conv2d(1, 32, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.ReLU(),
            nn.Conv2d(32, 1, 3, padding=1)
        )

        self.time_proj = nn.Linear(16, 1)

    def forward(self, x, t):
        time_info = self.time_proj(
            self.time_embed(t)
        ).view(-1, 1, 1, 1)

        return self.net(x + time_info)

model = NoisePredictor()
print("모델 준비 완료!")
```

```python
import torch
import torch.nn as nn

class ImprovedNoisePredictor(nn.Module):
    def __init__(self):
        super().__init__()

        # 시간 t를 8차원 벡터로 변환
        self.time_embed = nn.Embedding(T, 8)

        # 시간 정보를 CNN의 채널 수(32)에 맞춤
        self.time_proj = nn.Linear(8, 32)

        # 첫 번째 CNN
        self.conv1 = nn.Conv2d(1, 32, kernel_size=3, padding=1)

        # 두 번째 CNN
        self.conv2 = nn.Conv2d(32, 32, kernel_size=3, padding=1)

        # 예측한 노이즈 출력
        self.conv3 = nn.Conv2d(32, 1, kernel_size=3, padding=1)

        self.relu = nn.ReLU()

    def forward(self, x, t):
        # 시간 t를 벡터로 변환
        time_info = self.time_embed(t)

        # 8차원 -> 32차원
        time_info = self.time_proj(time_info)

        # 이미지의 높이와 너비 방향으로 차원을 추가
        time_info = time_info.view(-1, 32, 1, 1)

        # 첫 번째 CNN
        x = self.relu(self.conv1(x))

        # CNN 중간에 시간 정보 추가
        x = x + time_info

        # 두 번째 CNN
        x = self.relu(self.conv2(x))

        # 노이즈 예측
        x = self.conv3(x)

        return x
```

```python
import torch
import matplotlib.pyplot as plt

# 역방향 과정에 필요한 값
alpha_bar_prev = torch.cat([
    torch.tensor([1.0]),
    alphas_cumprod[:-1]
])

posterior_variance = (
    betas * (1 - alpha_bar_prev)
    / (1 - alphas_cumprod)
)

def generate_images(model, initial_noise):
    model.eval()

    # 두 모델에 동일한 초기 노이즈 사용
    x = initial_noise.clone()

    with torch.no_grad():
        for t in reversed(range(T)):
            t_batch = torch.full(
                (x.shape[0],), t, dtype=torch.long
            )

            predicted_noise = model(x, t_batch)

            beta_t = betas[t]
            alpha_t = alphas[t]
            alpha_bar_t = alphas_cumprod[t]

            # 역방향 과정: 노이즈 제거
            mean = (
                x - beta_t / torch.sqrt(1 - alpha_bar_t)
                * predicted_noise
            ) / torch.sqrt(alpha_t)

            if t > 0:
                x = mean + torch.sqrt(
                    posterior_variance[t]
                ) * torch.randn_like(x)
            else:
                x = mean

    return x.squeeze(1).cpu().numpy()
```

```python
# 두 모델이 같은 노이즈에서 시작하도록 설정
torch.manual_seed(42)
initial_noise = torch.randn(10, 1, 8, 8)

# 이미지 생성
samples_original = generate_images(model, initial_noise)
samples_improved = generate_images(
    model_improved, initial_noise
)

# 결과 비교
fig, axes = plt.subplots(2, 10, figsize=(15, 4))

for i in range(10):
    axes[0, i].imshow(
        samples_original[i], cmap="gray",
        vmin=0, vmax=1
    )
    axes[0, i].axis("off")

    axes[1, i].imshow(
        samples_improved[i], cmap="gray",
        vmin=0, vmax=1
    )
    axes[1, i].axis("off")

axes[0, 0].set_ylabel("Original", fontsize=12)
axes[1, 0].set_ylabel("Improved", fontsize=12)

plt.suptitle("DDPM: Original vs Improved Model")
plt.tight_layout()
plt.show()
```

```python
# 기존 개선 모델을 이어서 학습
for epoch in range(50):
    model_improved.train()
    total_loss = 0

    for (x0,) in loader:
        batch_size = x0.shape[0]

        # 무작위 시간 t 선택
        t = torch.randint(0, T, (batch_size,))

        # 실제 노이즈 생성
        noise = torch.randn_like(x0)

        # 정방향 과정: 노이즈 추가
        alpha_bar = alphas_cumprod[t].view(
            batch_size, 1, 1, 1
        )

        x_t = (
            torch.sqrt(alpha_bar) * x0
            + torch.sqrt(1 - alpha_bar) * noise
        )

        # 노이즈 예측
        predicted_noise = model_improved(x_t, t)

        # 실제 노이즈와 예측 노이즈의 차이
        loss = nn.functional.mse_loss(
            predicted_noise, noise
        )

        optimizer_improved.zero_grad()
        loss.backward()
        optimizer_improved.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(loader)
    loss_history_improved.append(avg_loss)

    if (epoch + 1) % 10 == 0:
        print(
            f"추가 학습 {epoch+1}/50 | "
            f"Loss: {avg_loss:.4f}"
        )
```
