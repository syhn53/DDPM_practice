```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision
import matplotlib.pyplot as plt
import time

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("PyTorch:", torch.__version__)
print("실행 장치:", device)

if device.type == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

train_dataset = datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

train_loader = DataLoader(
    train_dataset,
    batch_size=256,
    shuffle=True,
    num_workers=2,
    pin_memory=(device.type == "cuda")
)

print("학습 이미지 수:", len(train_dataset))
print("배치 수:", len(train_loader))
```

```python
images, labels = next(iter(train_loader))

fig, axes = plt.subplots(2, 5, figsize=(10, 4))

for i, ax in enumerate(axes.flat):
    img = images[i, 0] * 0.5 + 0.5
    ax.imshow(img, cmap="gray")
    ax.set_title(f"Label: {labels[i].item()}")
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
T = 200

betas = torch.linspace(
    1e-4, 0.02, T, device=device
)

alphas = 1.0 - betas

alpha_bars = torch.cumprod(
    alphas, dim=0
)

print("노이즈 단계:", T)
```

```python
def add_noise(x0, t):
    alpha_bar_t = alpha_bars[t].view(-1, 1, 1, 1)

    noise = torch.randn_like(x0)

    xt = (
        torch.sqrt(alpha_bar_t) * x0
        + torch.sqrt(1 - alpha_bar_t) * noise
    )

    return xt, noise
```

```python
x0 = images[:5].to(device)

timesteps = torch.tensor(
    [0, 30, 70, 120, 199],
    device=device
)

xt, noise = add_noise(x0, timesteps)

fig, axes = plt.subplots(2, 5, figsize=(12, 5))

for i in range(5):
    axes[0, i].imshow(
        x0[i, 0].cpu() * 0.5 + 0.5,
        cmap="gray"
    )
    axes[0, i].set_title("Original")
    axes[0, i].axis("off")

    axes[1, i].imshow(
        xt[i, 0].cpu().clamp(-1, 1) * 0.5 + 0.5,
        cmap="gray"
    )
    axes[1, i].set_title(f"t={timesteps[i].item()}")
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()
```

```python

# 1. U-Net 신경망 정의

class SmallUNet(nn.Module):
    def __init__(self, T):
        super().__init__()

        # 시간 단계 t를 벡터로 변환
        self.time_embedding = nn.Embedding(T, 32)
        self.time_fc = nn.Linear(32, 64)

        # 인코더: 이미지 특징 추출
        self.enc1 = nn.Conv2d(1, 32, 3, padding=1)
        self.enc2 = nn.Conv2d(
            32, 64, 3, stride=2, padding=1
        )

        # 중간 특징 처리
        self.mid = nn.Conv2d(64, 64, 3, padding=1)

        # 디코더: 특징 복원
        self.dec1 = nn.Conv2d(
            96, 32, 3, padding=1
        )

        # 최종 노이즈 예측
        self.out = nn.Conv2d(32, 1, 3, padding=1)

    def forward(self, x, t):
        # 시간 정보 생성
        temb = self.time_fc(
            self.time_embedding(t)
        )[:, :, None, None]

        # 인코더
        h1 = F.relu(self.enc1(x))
        h2 = F.relu(self.enc2(h1))

        # 시간 정보 반영
        h = F.relu(self.mid(h2 + temb))

        # 이미지 크기 복원
        h = F.interpolate(
            h, size=h1.shape[-2:], mode="nearest"
        )

        # Skip Connection
        h = torch.cat([h, h1], dim=1)

        # 디코더 및 노이즈 예측
        h = F.relu(self.dec1(h))
        return self.out(h)


# 2. 모델 생성
model = SmallUNet(T).to(device)

print("U-Net 생성 완료!")
```

```python

# 이미지 4장으로 테스트
x_test = images[:4].to(device)

# 각 이미지에 시간 단계 지정
t_test = torch.tensor(
    [10, 50, 100, 199],
    device=device
)

# U-Net에 입력
predicted_noise = model(x_test, t_test)

print("입력 이미지 크기:", x_test.shape)
print("예측 노이즈 크기:", predicted_noise.shape)
```

```python

# 모델 학습 설정
model = SmallUNet(T).to(device)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=1e-3
)

epochs = 5

# 학습 시작
for epoch in range(epochs):
    model.train()
    total_loss = 0

    for images, labels in train_loader:
        images = images.to(device)

        # 이미지마다 무작위 시간 단계 선택
        t = torch.randint(
            0, T, (images.shape[0],), device=device
        )

        # 정방향 과정: 이미지에 노이즈 추가
        xt, noise = add_noise(images, t)

        # U-Net이 노이즈 예측
        predicted_noise = model(xt, t)

        # 실제 노이즈와 예측 노이즈 비교
        loss = F.mse_loss(predicted_noise, noise)

        # 가중치 업데이트
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()

    avg_loss = total_loss / len(train_loader)

    print(
        f"Epoch [{epoch+1}/{epochs}], "
        f"Loss: {avg_loss:.4f}"
    )
```

```python

@torch.no_grad()
def generate_ddpm(num_samples=10):
    model.eval()

    # 완전한 무작위 노이즈에서 시작
    x = torch.randn(
        num_samples, 1, 28, 28, device=device
    )

    # t=199부터 t=0까지 역순으로 진행
    for t in reversed(range(T)):
        t_batch = torch.full(
            (num_samples,),
            t,
            device=device,
            dtype=torch.long
        )

        # 현재 이미지에 섞인 노이즈 예측
        predicted_noise = model(x, t_batch)

        alpha = alphas[t]
        alpha_bar = alpha_bars[t]
        beta = betas[t]

        # DDPM 역방향 평균 계산
        mean = (
            x - beta / torch.sqrt(1 - alpha_bar)
            * predicted_noise
        ) / torch.sqrt(alpha)

        if t > 0:
            # 역방향 과정의 분산
            alpha_bar_prev = alpha_bars[t - 1]

            variance = (
                beta * (1 - alpha_bar_prev)
                / (1 - alpha_bar)
            )

            # 다음 단계로 이동하며 노이즈 추가
            x = mean + torch.sqrt(variance) * torch.randn_like(x)
        else:
            # 마지막 단계에서는 추가 노이즈 없이 마무리
            x = mean

    return x.clamp(-1, 1)


# 숫자 이미지 10장 생성
start = time.time()
samples = generate_ddpm(10)

if device.type == "cuda":
    torch.cuda.synchronize()

print("생성 시간:", round(time.time() - start, 2), "초")

# 생성 결과 출력
fig, axes = plt.subplots(2, 5, figsize=(10, 4))

for i, ax in enumerate(axes.flat):
    img = samples[i, 0].cpu() * 0.5 + 0.5
    ax.imshow(img, cmap="gray")
    ax.axis("off")

plt.tight_layout()
plt.show()
```
